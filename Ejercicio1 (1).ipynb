```python
saldo = float(input("Ingrese el saldo inicial: "))
deposito = float(input("Ingrese el deposito: "))
retiro = float(input("Ingrese el retiro: "))

interes = 0.02

saldo = saldo + deposito
saldo = saldo - retiro

saldo_final = saldo + saldo * interes

print(f"Saldo final con interés: ${saldo_final:.2f}")
```

```python
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]
print("Lista inicial:", temperaturas)

temperaturas.append(23.7)
temperaturas.pop(0)
temperaturas.insert(1, 24.0)
temperaturas.sort()
print("Ordenadas:", temperaturas)

mayor = max(temperaturas)
menor = min(temperaturas)
diferencia = mayor - menor
suma = 0
for temperatura in temperaturas:
    suma = suma + temperatura
promedio = suma / len(temperaturas)
print("Diferencia:", diferencia)
print("Promedio:", promedio)

if 23.7 in temperaturas:
    print("Índice de 23.7 (comienza en 0):", temperaturas.index(23.7))
else:
    print("La temperatura 23.7 no está en la lista")

print("De la 2.ª a la 4.ª posición:", temperaturas[1:4])
print("Las tres primeras:", temperaturas[:3])
```

```python
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10}
]

print("Primer producto:", inventario[0]["nombre"])
inventario.append({"nombre": "Arroz", "precio": 4200, "stock": 8})

for producto in inventario:
    if producto["nombre"] == "Leche":
        producto["stock"] = producto["stock"] - 2

valor_total = 0
for producto in inventario:
    valor_total = valor_total + producto["precio"] * producto["stock"]
print("Valor total antes del aumento:", valor_total)

print("Productos con stock menor a 10:")
for producto in inventario:
    if producto["stock"] < 10:
        print(producto)

for producto in inventario:
    producto["precio"] = round(producto["precio"] * 1.10, 2)

buscado = input("Nombre del producto a buscar: ").strip().lower()
encontrado = False
for producto in inventario:
    if producto["nombre"].lower() == buscado:
        print("Producto encontrado:", producto)
        encontrado = True
        break
if not encontrado:
    print("Producto no encontrado")

indice = 0
while indice < len(inventario):
    if inventario[indice]["stock"] == 0:
        inventario.pop(indice)
    else:
        indice = indice + 1

# Ordenamiento imperativo por selección (sin lambda ni sorted con clave)
for i in range(len(inventario)):
    menor_indice = i
    for j in range(i + 1, len(inventario)):
        if inventario[j]["precio"] < inventario[menor_indice]["precio"]:
            menor_indice = j
    inventario[i], inventario[menor_indice] = inventario[menor_indice], inventario[i]

print("Inventario ordenado por precio:")
for producto in inventario:
    print(producto)
```

```python
punto = (3, 7)
print("Punto:", punto)
ancho = 5
alto = 3
rectangulo = (ancho, alto)
area = rectangulo[0] * rectangulo[1]
perimetro = 2 * (rectangulo[0] + rectangulo[1])
print("Área:", area)
print("Perímetro:", perimetro)

try:
    rectangulo[0] = 8
except TypeError as error:
    print("Error al modificar tupla:", error)

posiciones = [(0, 0), (1, 2), (3, 4)]
distancia_total = 0
for i in range(1, len(posiciones)):
    x1, y1 = posiciones[i - 1]
    x2, y2 = posiciones[i]
    distancia = ((x2 - x1)**2 + (y2 - y1)**2)**0.5
    distancia_total = distancia_total + distancia
print(f"Distancia total: {distancia_total:.2f}")
```

```python
biblioteca = {
    "libros": [
        {"titulo": "1984", "autor": "George Orwell", "prestado": False},
        {"titulo": "El Principito", "autor": "Antoine de Saint-Exupéry", "prestado": True}
    ],
    "usuarios": ["Ana", "Luis", "María"]
}

prestados = 0
disponibles = 0
for libro in biblioteca["libros"]:
    if libro["prestado"]:
        prestados = prestados + 1
    else:
        disponibles = disponibles + 1
print("Prestados al inicio:", prestados)
print("Disponibles al inicio:", disponibles)

biblioteca["libros"].append({"titulo": "Cien años de soledad", "autor": "Gabriel García Márquez", "prestado": False})
for libro in biblioteca["libros"]:
    if libro["titulo"] == "1984":
        libro["prestado"] = True

print("Biblioteca actualizada:")
print(biblioteca)
```
