```python
import requests

url = "https://jsonplaceholder.typicode.com/users"
try:
    respuesta = requests.get(url, timeout=15)
    respuesta.raise_for_status()
    usuarios = respuesta.json()
    nombres = []
    for usuario in usuarios:
        if usuario["email"].lower().endswith(".biz"):
            nombres.append(usuario["name"])
    print("Usuarios cuyo correo termina en .biz:")
    for nombre in nombres:
        print("-", nombre)
    if len(nombres) == 0:
        print("No se encontraron usuarios")
except requests.RequestException as error:
    print("No fue posible consultar la API:", error)
```

```python
import requests

nombre_pokemon = input("Ingrese el nombre de un Pokémon: ").strip().lower()
url = f"https://pokeapi.co/api/v2/pokemon/{nombre_pokemon}"
try:
    respuesta = requests.get(url, timeout=15)
    if respuesta.status_code == 200:
        datos = respuesta.json()
        tipos = []
        for elemento in datos["types"]:
            tipos.append(elemento["type"]["name"])
        habilidades = []
        for elemento in datos["abilities"]:
            habilidades.append(elemento["ability"]["name"])

        pokemon = {
            "nombre": datos["name"],
            "altura": datos["height"],
            "peso": datos["weight"],
            "tipos": tipos,
            "habilidades": habilidades
        }
        print("\nInformación del Pokémon")
        print("Nombre:", pokemon["nombre"].capitalize())
        print("Altura:", pokemon["altura"])
        print("Peso:", pokemon["peso"])
        print("Tipos:", ", ".join(pokemon["tipos"]))
        if len(tipos) > 1:
            print("Pokémon de tipo mixto")
        elif len(tipos) == 1:
            print("Pokémon de un solo tipo")
        else:
            print("Sin tipos registrados")
        if pokemon["peso"] > 100:
            print("Pokémon de gran tamaño")
        else:
            print("Pokémon ligero")
        print("Habilidades:")
        for habilidad in habilidades:
            print("-", habilidad)
        # Dato extra de la API: experiencia base
        print("Experiencia base:", datos["base_experience"])
    elif respuesta.status_code == 404:
        print("Pokémon no encontrado. Verifique el nombre ingresado.")
    else:
        print("Error del servidor. Código:", respuesta.status_code)
except requests.RequestException as error:
    print("Error de conexión:", error)
```
