```python
contador_global = 0

def incrementar_con_error():
    contador_global = contador_global + 1

try:
    incrementar_con_error()
except UnboundLocalError as error:
    print("Error original:", error)

contador_global = 0
def incrementar_global():
    global contador_global
    contador_global = contador_global + 1

incrementar_global()
print("Con global:", contador_global)

def incrementar_retorno(contador):
    contador = contador + 1
    return contador

contador = 0
contador = incrementar_retorno(contador)
print("Con parámetro y retorno:", contador)
```

```python
def agregar_nota(notas, nueva):
    print("id dentro de agregar antes:", id(notas))
    notas.append(nueva)

def reiniciar_original(notas):
    print("id dentro de reiniciar_original antes:", id(notas))
    notas = []
    print("id dentro de reiniciar_original después:", id(notas))

def reiniciar_correcto(notas):
    print("id dentro de reiniciar_correcto antes:", id(notas))
    notas.clear()
    print("id dentro de reiniciar_correcto después:", id(notas))

mis_notas = [3.5, 4.0]
print("id de mis_notas al inicio:", id(mis_notas))
agregar_nota(mis_notas, 4.8)
reiniciar_original(mis_notas)
print("Predicción y resultado con reiniciar_original:", mis_notas)
print("id después de la reasignación local:", id(mis_notas))
reiniciar_correcto(mis_notas)
print("Después de reiniciar_correcto:", mis_notas)
print("id final:", id(mis_notas))
```

```python
cuenta = {"titular": "Ana", "saldo": 200000, "movimientos": []}

def consultar(cuenta):
    print(f"Titular: {cuenta['titular']} - Saldo: ${cuenta['saldo']:.2f}")
    cuenta["movimientos"].append({"tipo": "consulta", "monto": 0, "saldo": cuenta["saldo"]})

def depositar(cuenta, monto):
    if monto <= 0:
        print("El depósito debe ser mayor que cero")
        return
    cuenta["saldo"] = cuenta["saldo"] + monto
    cuenta["movimientos"].append({"tipo": "depósito", "monto": monto, "saldo": cuenta["saldo"]})
    print("Depósito exitoso")

def retirar(cuenta, monto):
    if monto <= 0:
        print("El retiro debe ser mayor que cero")
    elif monto > cuenta["saldo"]:
        print("Saldo insuficiente. No se modificó la cuenta")
    else:
        cuenta["saldo"] = cuenta["saldo"] - monto
        cuenta["movimientos"].append({"tipo": "retiro", "monto": monto, "saldo": cuenta["saldo"]})
        print("Retiro exitoso")

while True:
    print("\n1. Consultar  2. Depositar  3. Retirar  4. Salir")
    opcion = input("Seleccione una opción: ").strip().lower()
    if opcion == "4" or opcion == "salir":
        break
    if opcion == "1":
        consultar(cuenta)
    elif opcion == "2" or opcion == "3":
        try:
            monto = float(input("Monto: "))
        except ValueError:
            print("Ingrese un número válido")
            continue
        if opcion == "2":
            depositar(cuenta, monto)
        else:
            retirar(cuenta, monto)
    else:
        print("Opción inválida")

print("\nEXTRACTO DE MOVIMIENTOS")
print(f"{'Tipo':<12}{'Monto':>14}{'Saldo':>16}")
print("-" * 42)
for movimiento in cuenta["movimientos"]:
    print(f"{movimiento['tipo']:<12}{movimiento['monto']:>14.2f}{movimiento['saldo']:>16.2f}")
print(f"Saldo final: ${cuenta['saldo']:.2f}")
```
