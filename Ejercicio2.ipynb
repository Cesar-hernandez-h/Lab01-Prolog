```python
edad = int(input("Ingrese su edad: "))
tiene_licencia = input("¿Tiene licencia? (s/n): ").strip().lower()
sancionado = input("¿Está sancionado? (s/n): ").strip().lower()

if edad >= 18 and tiene_licencia == "s" and sancionado == "n":
    print("Puede conducir")
elif edad < 18:
    print("No puede conducir: no cumple el requisito de edad de este ejercicio")
elif tiene_licencia != "s":
    print("No puede conducir: no tiene licencia")
else:
    print("No puede conducir: está sancionado o la respuesta es inválida")
```

```python
for i in range(1, 6):
    print(f"Iteración número {i}")

n = int(input("Ingrese n (número natural): "))
suma_naturales = 0
if n < 1:
    print("Ingrese un número natural mayor o igual a 1")
else:
    for i in range(1, n + 1):
        suma_naturales = suma_naturales + i
        if i % 2 == 0:
            print("Par:", i)
    print("Suma de los primeros n naturales:", suma_naturales)

temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]
suma = 0
for temperatura in temperaturas:
    suma = suma + temperatura
promedio = suma / len(temperaturas)
print("Promedio de temperaturas:", promedio)
print("Temperaturas mayores al promedio:")
for temperatura in temperaturas:
    if temperatura > promedio:
        print(temperatura)
```

```python
suma_pares = 0
suma_impares = 0
for i in range(1, 11):
    if i % 2 == 0:
        suma_pares = suma_pares + i
    else:
        suma_impares = suma_impares + i

estado_final = {"suma_pares": suma_pares, "suma_impares": suma_impares}
print(estado_final)
```

```python
productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ").strip()
    if nombre.lower() == "fin":
        break
    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))
    if precio < 0 or cantidad < 0:
        print("Precio o cantidad negativos: producto omitido")
        continue
    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

total = 0
for p in productos:
    total = total + p["precio"] * p["cantidad"]
print("Total del inventario:", total)

if len(productos) > 0:
    precios = []
    for p in productos:
        precios.append(p["precio"])
    precio_maximo = max(precios)
    for p in productos:
        if p["precio"] == precio_maximo:
            print("Producto más caro:", p)
            break
else:
    print("El inventario está vacío")
print("Estado final:", productos)
```
